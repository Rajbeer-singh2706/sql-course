## Step 6 — Run Your Very First SQL Query

Let's ask PostgreSQL what version it is running.  
This is the "Hello, World!" of database connections.

> The `%%sql` magic (double `%`) means the **entire cell** is sent to the database as SQL.


````python
%%sql

SELECT version();
````

## Step 7 — Explore Your Database

Let's find out which database we are connected to and check the current timestamp.
These are quick diagnostic queries — useful any time you want to confirm your connection.


````python
%%sql

-- Which database am I connected to?
SELECT current_database() AS database_name,
       current_user       AS logged_in_as,
       NOW()              AS server_time;
````

## Step 8 — List Tables in Your Database

In psql you would type `\dt` to list tables.  
In SQL we query the **information schema** — a built-in set of views that describe the database structure.


````python
%%sql

-- List all tables in the public schema
-- (this is the SQL equivalent of psql's \dt command)
SELECT table_name,
       table_type
FROM   information_schema.tables
WHERE  table_schema = 'public'
ORDER  BY table_name;
````

> **Expected output:** If your database is brand new, you will see an empty result — that is perfectly normal!  
> We will start creating tables from **Day 23** of the course.  
> For now, all our data will be created inline in queries using `VALUES`.


## Step 9 — Check PostgreSQL Version Details

Let's confirm we are on PostgreSQL 16 (or close to it).


````python
%%sql

SELECT current_setting('server_version')        AS pg_version,
       current_setting('server_version_num')    AS pg_version_num;
````

## Step 10 — Your First Real Query (with Inline Data)

We don't have tables yet — but PostgreSQL lets us create data on the fly using `VALUES`.  
This is a sneak peek at what Day 2 will cover in full.

Run this to prove everything is working end-to-end:


````python
%%sql

-- Create a temporary result set using VALUES (no table needed!)
-- This is a preview of what we will learn on Day 2
SELECT *
FROM (
    VALUES
        (1, 'Alice',   'Data Engineer'),
        (2, 'Bob',     'Analytics Engineer'),
        (3, 'Charlie', 'Data Scientist')
) AS team(id, name, role);
````
